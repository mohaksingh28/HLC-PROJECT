```python
class Item:
    def __init__(self, name, qty, price, min_qty=1):
        self.name, self.qty, self.price, self.min_qty = name, qty, price, min_qty
class Inventory:
    def __init__(self): self.items = {}
    def add_item(self, name, qty, price, min_qty=1):
        self.items[name] = Item(name, qty, price, min_qty); print(f"Added {name}")
    def update_quantity(self, name, qty):
        if name in self.items: self.items[name].qty = qty
        else: print("Item not found")
    def show_inventory(self):
        for i in self.items.values():
            flag = "  <- REPLACE" if i.qty <= i.min_qty else ""
            print(f"{i.name:<10} qty:{i.qty:<4} Rs.{i.price}{flag}")
inv = Inventory()
inv.add_item("Rice", 5, 60, 2)
inv.add_item("Soap", 1, 35)
inv.update_quantity("Rice", 2)
inv.show_inventory()
```
